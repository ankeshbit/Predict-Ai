# 00 - Predict-Ai Environment Verification
This notebook verifies that Google Colab runtime meets the dependency requirements and library version pins defined in ml/constraints.txt.

```python
import platform
import sys

print('=== Python & OS Runtime ===')
print('Python version:', sys.version)
print('Platform:', platform.platform())
```

```python
packages = ['numpy', 'pandas', 'scipy', 'sklearn', 'xgboost', 'shap', 'joblib', 'yaml', 'pydantic']

print('=== Package Version Check ===')
for pkg in packages:
    try:
        mod = __import__(pkg)
        ver = getattr(mod, '__version__', 'unknown')
        print(f'{pkg}: {ver}')
    except ImportError as e:
        print(f'{pkg}: NOT INSTALLED ({e})')
```
