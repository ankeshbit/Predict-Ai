```python
# Constant Sensor Channel Identification
if 'df' in locals():
    stds = df.drop(columns=['unit_id', 'cycle']).std()
    low_var = stds[stds < 0.001].index.tolist()
    print('Near-zero variance channels dropped in training:', low_var)
```
